```python
import xarray as xr
from pathlib import Path
```

```python
uy_dir = Path("../data/time_series")
```

```python
uy_files = sorted(uy_dir.glob("CAMELS_UY_*.nc"))
uy_files
```

```python
for f in uy_files:
    ds = xr.load_dataset(f)
    ds_renamed = ds.rename({'prcp_mm_day': 'total_precipitation_sum'})
    ds_renamed.to_netcdf(Path.cwd() / f.name)  # saves to current working directory
    print(f"Done: {f.name}")

print("All files updated!")
```
