```python
from neuralhydrology.nh_run import start_run, eval_run, finetune
from pathlib import Path
import pandas as pd
import pickle
import numpy as np
import yaml
```

```python
RUN_DIR = Path("./runs")
```

```python
# Find all directories starting with 'precip'
precip_dirs = [d for d in RUN_DIR.iterdir() if d.is_dir()] #and d.name.startswith("chirps")]

print(f"Found {len(precip_dirs)} directories: {[d.name for d in precip_dirs]}")
```

```python
run_dir=precip_dirs[0]
run_dir
```

```python
config_path = run_dir / "config.yml"

# Read the config
with open(config_path, 'r') as f:
    content = f.read()

if "test_start_date" not in content:
    content += "\ntest_start_date: 01/10/2008"
    print(f"  Added test_start_date")
else:
    print(f"  test_start_date already present, skipping")

if "test_end_date" not in content:
    content += "\ntest_end_date: 30/09/2014"
    print(f"  Added test_end_date")
else:
    print(f"  test_end_date already present, skipping")

# Fix data_dir path
content = content.replace(
    "data_dir: ../extended_dataset/data",
    "data_dir: ../../extended_dataset/data"
)

# Write once at the end with all changes
with open(config_path, 'w') as f:
    f.write(content)

# Run evaluation
print(f"  Running eval_run...")
eval_run(run_dir=run_dir, period="test")
print(f"  Done!")
```
