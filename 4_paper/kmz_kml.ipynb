```python
import pandas as pd
import numpy as np
from pathlib import Path
import simplekml
```

```python
# --------------- Paths ------------------------
area_file= Path("/inputs/data_updated_2/attributes/attributes_other.csv")
coordenadas_dir = Path("/home/azureuser/NeuralHydrologyAzure/1_preparing_data/1_uy_basins_data/gauge_data/precip_uy/COORDENADAS.xlsx")
```

# streamflow stations kml (UY)

```python
area = pd.read_csv(area_file)
area
```

```python
keep_ids = [
    "CAMELS_UY_2", "CAMELS_UY_3", "CAMELS_UY_5", "CAMELS_UY_6",
    "CAMELS_UY_7", "CAMELS_UY_8", "CAMELS_UY_9", "CAMELS_UY_10",
    "CAMELS_UY_11", "CAMELS_UY_15", "CAMELS_UY_16",
]

df_sub = (
    area[area["gauge_id"].isin(keep_ids)]
    [["gauge_name", "gauge_lat", "gauge_lon"]]
    .reset_index(drop=True)
)

print(df_sub)
assert len(df_sub) == len(keep_ids), "Some IDs weren't found"
```

```python
kml = simplekml.Kml()
for row in df_sub.itertuples():
    kml.newpoint(name=row.gauge_name, coords=[(row.gauge_lon, row.gauge_lat)])

kml.save("uruguay_gauges.kml")
```

# precipitation stations kml (UY)

```python
coordenadas = pd.read_excel(coordenadas_dir)
coordenadas
```

```python
estaciones = ['Capilla del Sauce', 'Canada Grande', 'Cuchilla Caraguata Norte',
       'Cuchilla del Ombu', 'El Molino', 'Goni', 'Minas de Corrales',
       'Moirones', 'Palmitas', 'Paso de la Cruz_rn', 'Polanco del Yi',
       'Rossell y Rius', 'Tranqueras', 'Vichadero', 'Rivera',
       'Cerro de las Cuentas', 'Cerro Chato']
```

```python
coordenadas_filtrado = coordenadas.loc[
    coordenadas["Estacion"].isin(estaciones), ["Estacion", "lat", "lon"]
].reset_index(drop=True)

coordenadas_filtrado
```

```python
kml = simplekml.Kml()
for row in coordenadas_filtrado.itertuples():
    kml.newpoint(name=row.Estacion, coords=[(row.lon, row.lat)])

kml.save("uruguay_precip_gauges.kml")
```
