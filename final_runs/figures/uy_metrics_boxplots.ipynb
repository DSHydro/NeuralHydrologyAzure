```python
from pathlib import Path
import pandas as pd
import torch
import matplotlib.pyplot as plt
import xarray as xr
import yaml
import geopandas as gpd
import numpy as np
import pickle
import itertools
```

```python
# --------------------- Paths -------------------
ensemble_metrics_dir = Path("../all_ensemble_metrics")
```

```python
# -------------------- NSE data for Uruguay basins (runs just with UY basins) -------------------
precip_products = [
    "prcp_mm_day",
    "prcp_chirps_mm_day",
    "prcp_mswep_mm_day",
    "prcp_gauge_mm_day",
]

all_combinations = [
    combo
    for r in range(1, len(precip_products) + 1)
    for combo in itertools.combinations(precip_products, r)
]

# all_combinations

positions = [0, 1, 2, 3, 7, 13]  # whichever positions you want
all_combinations = [all_combinations[i] for i in positions]

all_combinations
```

```python
metric_names = [
    'FHV', 'FMS', 'FLV',
    'Peak-Timing', 'Missed-Peaks', 'Peak-MAPE'
]
```

```python
# Dict of {metric: list of arrays (one per combo)}
all_metrics = {m: [] for m in metric_names}
model_labels = {}
basins_order = []

for i, combo in enumerate(all_combinations, start=1):
    precip_name = "_".join(combo)
    metrics_csv = pd.read_csv(ensemble_metrics_dir / f"precip_{precip_name}.csv")

    if i == 1:
        basins_order = metrics_csv["basin_id"].values

    for metric in metric_names:
        all_metrics[metric].append(metrics_csv[metric].values)

    model_labels[i] = precip_name
```

```python
dataset_labels = [
    "CARAVAN",
    "CHIRPS",
    "MSWEP",
    "GAUGE",
    # "CARAVAN\nCHIRPS",
    # "CARAVAN\nMSWEP",
    # "CARAVAN\nGAUGE",
    "CHIRPS\nMSWEP",
    # "CHIRPS\nGAUGE",
    # "MSWEP\nGAUGE",
    # "CARAVAN\nCHIRPS\nMSWEP",
    # "CARAVAN\nCHIRPS\nGAUGE",
    # "CARAVAN\nMSWEP\nGAUGE",
    "CHIRPS\nMSWEP\nGAUGE",
    # "CARAVAN\nCHIRPS\nMSWEP\nGAUGE"
]
```

```python
for metric in metric_names:
    fig, ax = plt.subplots(figsize=(14, 6))
    
    ax.boxplot(all_metrics[metric], labels=dataset_labels)
    ax.set_title(metric, fontsize=14)
    # ax.tick_params(axis='x', rotation=45)
    ax.set_ylabel(metric)
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    
    plt.tight_layout()
    # plt.savefig(ensemble_metrics_dir / f"boxplot_{metric}.png", dpi=150, bbox_inches='tight')
    plt.show()
```

```python
metric_limits = {
    'FHV':          (-20, 20),
    'FMS':          (-30, 30),
    'FLV':          (-50, 50),
    'Peak-Timing':  (0.5,   1.5),
    'Missed-Peaks': (0.4, 0.6),
    'Peak-MAPE':    (0, 60),
}

fig, axes = plt.subplots(len(metric_names), 1, figsize=(12, 15))

for ax, metric in zip(axes, metric_names):
    medians = np.array([np.nanmedian(values) for values in all_metrics[metric]])
    data_2d = medians.reshape(1, -1)
    
    vmin, vmax = metric_limits[metric]
    im = ax.imshow(data_2d, aspect='auto', cmap='viridis', vmin=vmin, vmax=vmax)
    
    plt.colorbar(im, ax=ax, orientation='vertical', fraction=0.02, pad=0.02)
    
    for j, val in enumerate(medians):
        ax.text(j, 0, f"{val:.2f}", ha='center', va='center', fontsize=8,
                color='white' if val < np.nanmax(medians) * 0.7 else 'black')
    
    ax.set_title(metric, fontsize=11, fontweight='bold')
    ax.set_yticks([])
    ax.set_xticks(range(len(dataset_labels)))
    ax.set_xticklabels(dataset_labels) #, fontsize=8, rotation=45, ha='right')

plt.tight_layout()
plt.savefig(ensemble_metrics_dir / "heatmap_median_metrics.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.colors as mcolors
import matplotlib.cm as mcm

metric_limits = {
    'FHV':          (-20, 20),
    'FMS':          (-30, 30),
    'FLV':          (-50, 50),
    'Peak-Timing':  (0.5,  1.5),
    'Missed-Peaks': (0.4,  0.6),
    'Peak-MAPE':    (0,    60),
}

# Define per-metric colormap and normalization
# "best" direction: for abs-error metrics, closer to 0 is best → diverging centered at 0
# For Peak-Timing, higher is best → green at top
# For Missed-Peaks and Peak-MAPE, lower is best → green at bottom
def make_cmap_and_norm(metric, vmin, vmax):
    if metric in ('FHV', 'FMS', 'FLV'):
        # Diverging: 0 is best (green center), extremes are red
        cmap = mcm.get_cmap('RdYlGn')
        norm = mcolors.TwoSlopeNorm(vmin=vmin, vcenter=0, vmax=vmax)
    else:
        # Lower is better (Peak-Timing, Missed-Peaks, Peak-MAPE) → green at bottom
        cmap = mcm.get_cmap('RdYlGn_r')
        norm = mcolors.Normalize(vmin=vmin, vmax=vmax)
    return cmap, norm

fig, axes = plt.subplots(len(metric_names), 1, figsize=(12, 15))

for ax, metric in zip(axes, metric_names):
    medians = np.array([np.nanmedian(values) for values in all_metrics[metric]])
    data_2d = medians.reshape(1, -1)

    vmin, vmax = metric_limits[metric]
    cmap, norm = make_cmap_and_norm(metric, vmin, vmax)

    im = ax.imshow(data_2d, aspect='auto', cmap=cmap, norm=norm)
    plt.colorbar(im, ax=ax, orientation='vertical', fraction=0.02, pad=0.02)

    for j, val in enumerate(medians):
        # Determine text color based on background brightness
        rgba = cmap(norm(val))
        brightness = 0.299 * rgba[0] + 0.587 * rgba[1] + 0.114 * rgba[2]
        text_color = 'black' if brightness > 0.5 else 'white'
        ax.text(j, 0, f"{val:.2f}", ha='center', va='center', fontsize=8, color=text_color)

    ax.set_title(metric, fontsize=11, fontweight='bold')
    ax.set_yticks([])
    ax.set_xticks(range(len(dataset_labels)))
    ax.set_xticklabels(dataset_labels)

plt.tight_layout()
# plt.savefig(ensemble_metrics_dir / "heatmap_median_metrics.png", dpi=150, bbox_inches='tight')
plt.show()
```
